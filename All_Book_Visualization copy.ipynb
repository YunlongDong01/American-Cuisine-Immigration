```python
import json
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import pandas as pd
import seaborn as sns
from collections import Counter
from itertools import combinations
from collections import defaultdict
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 中文显示配置
plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False

def load_and_analyze(metadata_path, recipes_path):
    """加载数据，数据集概览"""
    with open(metadata_path, "r", encoding="utf-8") as f:
        metadata = json.load(f)
    with open(recipes_path, "r", encoding="utf-8") as f:
        recipes = json.load(f)
    
    df_books = pd.DataFrame(metadata)
    df_recipes = pd.DataFrame(recipes)

    
    return df_books, df_recipes

df_books, df_recipes = load_and_analyze("cookbooks_extracted_metadata.json", "cookbooks_extracted_recipes.json")

print(f"书籍数量: {len(df_books)}")
print(f"菜谱数量: {len(df_recipes)}")
print(f"最早出版年份: {df_books['date'].min()}" if 'date' in df_books.columns else "")
print(f"最晚出版年份: {df_books['date'].max()}" if 'date' in df_books.columns else "")
print(f"民族分类数量: {df_books['ethnic_group'].nunique()}")
print(f"菜系类别数量: {df_books['class1'].nunique()}\n")

print('df_books的column: \n',df_books.columns)
print('\ndf_recipes的column: \n',df_recipes.columns)

# 正确查询特定book_id的年份
print(df_books[df_books['book_id'] == 'jenn']['date'])
```

```python
# 打印其中一个菜谱的详细信息
def print_recipe_details(recipes, books, index):
    """打印菜谱详细信息"""
    if index < len(recipes):
        recipe = recipes.iloc[index]
        print(f"菜谱名称: {recipe['name']}")
        
        # 通过 book_id 找书籍信息
        book_id = recipe['book_id']
        book = books[books['book_id'] == book_id]
        if not book.empty:
            book = book.iloc[0]
            print(f"书籍名称: {book['title']}")
            print(f"书籍作者: {book['author']}")
            print(f"书籍出版年份: {book['date']}")
        else:
            print("未找到对应书籍信息。")
        
        print(f"菜系类别: {recipe['class1']}")
        print(f"民族分类: {recipe['ethnic_group']}")
        print(f"菜谱材料: {recipe['ingredients']}")
        print(f"厨具工具: {recipe['implements']}")
        print(f"变体: {recipe['variations'] if 'variations' in recipe else '无'}")
        
    else:
        print("索引超出范围，请输入有效的索引。")

# 调用示例
print_recipe_details(df_recipes, df_books, 0)
```

```python
def clean_ethnic_group(s):
    
    """将 ethnic_group 字段标准化并分拆多个民族"""
    if not s or not isinstance(s, str):
        return []
    
    # 拆分多个民族，以逗号分隔，去除空格和统一小写再大写首字母
    groups = [g.strip().capitalize() for g in s.split(',') if g.strip()]
    return groups

# 应用到整个数据框：构建民族列表列
df_books['ethnic_group'] = df_books['ethnic_group'].replace('', np.nan)
df_books['ethnic_group'] = df_books['ethnic_group'].fillna('Unknown')
df_books['ethnic_group_list'] = df_books['ethnic_group'].apply(clean_ethnic_group)
# 展开所有民族，重新统计频数
from collections import Counter
ethnic_flat_list = [group for sublist in df_books['ethnic_group_list'] for group in sublist]
ethnic_counts = pd.Series(Counter(ethnic_flat_list)).sort_values(ascending=False).head(10)
```

```python
def preprocess_subjects(df):
    unwanted_categories = {"Cookery", "Recipes", "Menus", "Food", "Womencooks", 
                           "Confectionery", "Cookbooks", "Cookeryforthesick", "Liquors", 
                           "Vegetarianism", "Formula", "Medicine,Popular", "Chocolate", "Cocoa",
                           "Groceries", "Military", "Waiters", "Food--Dictionaries","Homeeconomics", 
                           "HomeEconomics","Diet","Juvenileliterature",
                           "Immigrants--UnitedStates"}
    
    def clean_subjects(subjects):
        cleaned = []
        for subject in subjects:
            if subject.startswith("Cookery,"):
                subject = subject.replace("Cookery,", "").strip()
            if subject.startswith("Cookery--"):
                subject = subject.replace("Cookery--", "").strip()
            if subject not in unwanted_categories:
                cleaned.append(subject)
        return cleaned

    df['subject'] = df['subject'].apply(clean_subjects)

def categorize_subjects(subjects):
    categorized = []
    seen = set()  # 用于跟踪已添加的分类
    
    for subject in subjects:
        # 处理American相关分类
        if (subject.startswith("American--") or 
            subject in {"Northeastern", "SouthernStates", "MiddleEastern",
                       "Washington", "California", "Quakercooking", "Virginia", 
                       "Kentucky", "NewEngland", "NewYork", "Pennsylvania", "Ohio"}):
            if "American" not in seen:
                categorized.append("American")
                seen.add("American")
        
        # 处理German相关分类
        elif subject.startswith("Germans--"):
            if "German" not in seen:
                categorized.append("German")
                seen.add("German")
        
        # 其他分类
        elif subject not in seen:  # 避免重复
            categorized.append(subject)
            seen.add(subject)
    
    return categorized
```

```python
# 创建 df_books 的副本，并预处理 subject 类别
df_books_copy = df_books.copy()
preprocess_subjects(df_books_copy)
df_books_copy['categorized_subject'] = df_books_copy['subject'].apply(categorize_subjects)

def visualize_dataset_overview(df_books, df_recipes):
    import matplotlib.pyplot as plt
    import seaborn as sns
    import pandas as pd
    import re
    # 图形风格设置
    sns.set(style="whitegrid")
    plt.rcParams['font.sans-serif'] = ['SimHei']
    plt.rcParams['axes.unicode_minus'] = False

    

    # 设置画布
    fig, axs = plt.subplots(3, 2, figsize=(18, 16))
    fig.suptitle("美国食谱数据集概览", fontsize=20)

    ### 1. 出版年份分布
    if 'date' in df_books.columns:
        def extract_year_simple(date_val):
            date_str = str(date_val)  # 强制转为字符串
            
            # 查找c后面的4位数字
            if 'c' in date_str.lower():  
                match = re.search(r'c\D*(\d{4})', date_str, re.IGNORECASE)
                if match:
                    # print(f"找到年份: {match.group(1)}")
                    return int(match.group(1))
            
            # 查找任何4位数字
            match = re.search(r'(\d{4})', date_str)
            if match:
                return int(match.group(1))
            
            return np.nan  
        
        df_books['date'] = df_books['date'].apply(extract_year_simple)
        
        valid_years = df_books['date'].dropna()
        if not valid_years.empty:
            sns.histplot(valid_years, bins=20, kde=True, ax=axs[0, 0])
            axs[0, 0].set_title(f"出版年份分布（共{len(valid_years)}条数据）")
        else:
            axs[0, 0].text(0.5, 0.5, "无有效年份数据", ha='center')

    ### 2. 民族分类数量（书籍）
    if 'ethnic_group' in df_books.columns:
        # 使用之前构造的标准化频数
        sns.barplot(x=ethnic_counts.values, y=ethnic_counts.index, ax=axs[0, 1], hue=ethnic_counts.index, legend=False, palette='viridis')
        axs[0, 1].set_title("Top 10 (标准化)民族分类书籍数量")
        axs[0, 1].set_xlabel("数量")
        axs[0, 1].set_ylabel("民族分类")
        for p in axs[0, 1].patches:
            axs[0, 1].text(p.get_width() + 1, p.get_y() + p.get_height() / 2, 
                           int(p.get_width()), va='center')


    ### 3. 菜系类别数量
    if 'class1' in df_books.columns:
        class_counts = df_books['class1'].value_counts().head(10)
        sns.barplot(x=class_counts.values, y=class_counts.index, ax=axs[1, 0], hue=class_counts.index, legend=False, palette='viridis')
        axs[1, 0].set_title("菜系类别书籍数量")
        axs[1, 0].set_xlabel("数量")
        axs[1, 0].set_ylabel("菜系类别")
        for p in axs[1, 0].patches:
            axs[1, 0].text(p.get_width() + 1, p.get_y() + p.get_height() / 2, 
                           int(p.get_width()), va='center')


    ### 4. Categorized Subject（书）
    unique_subjects = df_books_copy.explode('categorized_subject')[['book_id', 'categorized_subject']].drop_duplicates()
    subject_counts_books = unique_subjects['categorized_subject'].value_counts()
    sns.barplot(x=subject_counts_books.index, y=subject_counts_books.values, ax=axs[1, 1], 
                hue=subject_counts_books.index, legend=False, palette='viridis')
    axs[1, 1].set_title("书籍主题分布（唯一计数）")
    axs[1, 1].set_ylabel("数量")
    axs[1, 1].tick_params(axis='x', rotation=45)
    for p in axs[1, 1].patches:
        axs[1, 1].text(p.get_x() + p.get_width() / 2, p.get_height() + 1, 
                           int(p.get_height()), ha='center')


    ### 5. Categorized Subject（食谱）
    book_subjects_exploded = df_books_copy[['book_id', 'categorized_subject']].explode('categorized_subject')
    recipes_with_subjects = df_recipes.merge(book_subjects_exploded, on='book_id', how='left')
    subject_counts_recipes = recipes_with_subjects['categorized_subject'].value_counts()
    sns.barplot(x=subject_counts_recipes.index, y=subject_counts_recipes.values, ax=axs[2, 0], 
                hue=subject_counts_recipes.index, legend=False, palette='coolwarm')
    axs[2, 0].set_title("食谱主题分布")
    axs[2, 0].set_ylabel("数量")
    axs[2, 0].tick_params(axis='x', rotation=45)
    for p in axs[2, 0].patches:
        axs[2, 0].text(p.get_x() + p.get_width() / 2, p.get_height() + 1, 
                           int(p.get_height()), ha='center')


    ### 6. 民族分类（食谱）
    df_recipes['ethnic_group'] = df_recipes['ethnic_group'].replace('', pd.NA).fillna('Unknown')
    ethnic_counts_recipes = df_recipes['ethnic_group'].value_counts().head(12)
    sns.barplot(x=ethnic_counts_recipes.index, y=ethnic_counts_recipes.values, ax=axs[2, 1], 
                hue=ethnic_counts_recipes.index, legend=False, palette='mako')
    axs[2, 1].set_title("Top 12 民族分类（食谱）")
    axs[2, 1].set_ylabel("数量")
    axs[2, 1].tick_params(axis='x', rotation=45)
    for p in axs[2, 1].patches:
        axs[2, 1].text(p.get_x() + p.get_width() / 2, p.get_height() + 1, 
                           int(p.get_height()), ha='center')


    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()


# 调用函数
visualize_dataset_overview(df_books, df_recipes)
```

```python
def plot_top_books_by_year_ordered(df_recipes, df_books, top_n=74):
    """
    绘制按出版年份排序的每本书中菜谱数量分布（前 top_n 本书）
    横轴标签改为对应的出版年份
    """
    # 统计每本书的菜谱数量
    recipe_counts = df_recipes['book_id'].value_counts().reset_index()
    recipe_counts.columns = ['book_id', 'recipe_count']
    
    # 从df_books获取出版年份和书名
    books_with_info = df_books[['book_id', 'date', 'title']].copy()
    
    # 合并数据
    merged = recipe_counts.merge(books_with_info, on='book_id', how='left')
   
    # 转换为数值型年份，处理缺失值
    merged['date'] = pd.to_numeric(merged['date'], errors='coerce')
    merged = merged.dropna(subset=['date'])
    
    # 按年份排序
    merged = merged.sort_values('date').head(top_n)
    
    plt.figure(figsize=(12, 6))
    
    # 使用条形图位置索引（0到73），但标签改为年份: 将同一年的书籍放在一起
    x_positions = range(len(merged))
    plt.bar(x_positions, merged['recipe_count'], color='steelblue', width=0.8)
    
    # 设置x轴标签为年份
    plt.xticks(x_positions, merged['date'].astype(int), rotation=45)
    
    plt.title("Recipe Count per Cookbook (Ordered by Publication Year)", fontsize=14)
    plt.xlabel("Publication Year", fontsize=12)
    plt.ylabel("Number of Recipes", fontsize=12)
    
    plt.tight_layout()
    plt.show()

# 调用函数
plot_top_books_by_year_ordered(df_recipes, df_books, top_n=74)
```

```python
from wordcloud import WordCloud
from collections import Counter
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors

def generate_wordcloud_by_freq_color(df, max_words=30):
    # 统计频率
    all_ingredients = [
        ing
        for lst in df['ingredients']
        for ing in lst
    ]
    freq = Counter(all_ingredients)
    filtered_freq = dict(freq.most_common(max_words))

    # 创建椭圆遮罩
    h, w = 400, 800
    y, x = np.ogrid[:h, :w]
    cy, cx = h // 2, w // 2
    ry, rx = h // 2.2, w // 2.2
    mask = ((y - cy) / ry) ** 2 + ((x - cx) / rx) ** 2 > 1
    mask = 255 * mask.astype(np.uint8)

    # 自定义颜色函数
    max_freq = max(filtered_freq.values())
    cmap = plt.get_cmap("viridis_r")

    def freq_color_func(word, font_size, position, orientation, random_state=None, **kwargs):
        # Normalize freq to [0, 1]
        freq_norm = filtered_freq.get(word, 0) / max_freq
        r, g, b, _ = cmap(freq_norm)
        return f"rgb({int(r*255)}, {int(g*255)}, {int(b*255)})"

    # 创建词云
    wc = WordCloud(
        width=w,
        height=h,
        background_color="white",
        max_words=max_words,
        mask=mask,
        color_func=freq_color_func
    ).generate_from_frequencies(filtered_freq)

    # 展示图像
    plt.figure(figsize=(12, 6))
    plt.imshow(wc, interpolation="bilinear")
    plt.title("Word Cloud with Frequency-based Colors", fontsize=16)
    plt.axis("off")
    plt.tight_layout()
    plt.show()

generate_wordcloud_by_freq_color(df_recipes)
```

```python
# 中国类全食材图
def generate_ethnic_group_wordcloud(df, ethnic_group, max_words=30):
    """Generate a word cloud for recipes belonging to a specific ethnic group."""
    # 筛选出指定 ethnic_group 的食谱
    subset = df[df['ethnic_group'] == ethnic_group]

    # 如果没有匹配的菜系，就提示一下
    if subset.empty:
        print(f"No recipes found for ethnic group: {ethnic_group}")
        return

    # 收集所有食材
    all_ingredients = [
        ing
        for lst in subset['ingredients']
        for ing in lst
    ]

    # 计算频率并取前 max_words 个
    freq = Counter(all_ingredients)
    filtered_freq = dict(freq.most_common(max_words))

    # 创建椭圆遮罩
    h, w = 400, 800
    y, x = np.ogrid[:h, :w]
    cy, cx = h // 2, w // 2
    ry, rx = h // 2.2, w // 2.2
    mask = ((y - cy) / ry) ** 2 + ((x - cx) / rx) ** 2 > 1
    mask = 255 * mask.astype(np.uint8)

    # 自定义颜色函数
    max_freq = max(filtered_freq.values())
    cmap = plt.get_cmap("inferno_r")

    def freq_color_func(word, font_size, position, orientation, random_state=None, **kwargs):
        freq_norm = filtered_freq.get(word, 0) / max_freq
        r, g, b, _ = cmap(freq_norm)
        return f"rgb({int(r*255)}, {int(g*255)}, {int(b*255)})"

    # 创建词云对象
    wc = WordCloud(
        width=w,
        height=h,
        background_color="white",
        max_words=max_words,
        mask=mask,
        color_func=freq_color_func
    ).generate_from_frequencies(filtered_freq)
    # 打印前十食材
    print(f"Top ingredients for {ethnic_group}:")
    for i, (ing, count) in enumerate(freq.most_common(10), 1):
        print(f"{i}. {ing}: {count}")

    # 展示图像
    plt.figure(figsize=(12, 6))
    plt.imshow(wc, interpolation="bilinear")
    plt.title(f"{ethnic_group.capitalize()} Recipes Ingredient Word Cloud", fontsize=16)
    plt.axis("off")
    plt.tight_layout()
    plt.show()

# 示例调用
generate_ethnic_group_wordcloud(df_recipes, ethnic_group="chinese")
```

```python
# 中国类过滤通用食材图
def generate_filtered_ethnic_group_wordcloud(df, ethnic_group, max_words=30, common_words=None):
    """Generate a word cloud for a specific ethnic group, excluding common words, with frequency-based color and elliptical mask."""
    # 如果没有提供 common_words，则默认从全食谱中提取前20个常见词
    if common_words is None:
        all_ingredients = [ing for lst in df['ingredients'] for ing in lst]
        common_freq = Counter(all_ingredients)
        common_words = {word for word, _ in common_freq.most_common(100)}
        print(f"Common words (filtered out): {common_words}")

    # 筛选出指定民族菜系的食谱
    subset = df[df['ethnic_group'] == ethnic_group]
    if subset.empty:
        print(f"No recipes found for ethnic group: {ethnic_group}")
        return

    # 收集并过滤食材
    all_ingredients = [
        ing
        for lst in subset['ingredients']
        for ing in lst
        if ing not in common_words
    ]
    freq = Counter(all_ingredients)
    filtered_freq = dict(freq.most_common(max_words))

    # 打印前十食材
    print(f"\nTop ingredients for {ethnic_group} (excluding common words):")
    for i, (ing, count) in enumerate(freq.most_common(10), 1):
        print(f"{i}. {ing}: {count}")

    # 创建椭圆遮罩
    h, w = 400, 800
    y, x = np.ogrid[:h, :w]
    cy, cx = h // 2, w // 2
    ry, rx = h // 2.2, w // 2.2
    mask = ((y - cy) / ry) ** 2 + ((x - cx) / rx) ** 2 > 1
    mask = 255 * mask.astype(np.uint8)

    # 定义颜色函数
    max_freq = max(filtered_freq.values()) if filtered_freq else 1
    cmap = plt.get_cmap("inferno_r")

    def freq_color_func(word, font_size, position, orientation, random_state=None, **kwargs):
        freq_norm = filtered_freq.get(word, 0) / max_freq
        r, g, b, _ = cmap(freq_norm)
        return f"rgb({int(r*255)}, {int(g*255)}, {int(b*255)})"

    # 生成词云
    wc = WordCloud(
        width=w,
        height=h,
        background_color="white",
        max_words=max_words,
        mask=mask,
        color_func=freq_color_func
    ).generate_from_frequencies(filtered_freq)

    # 显示图像
    plt.figure(figsize=(12, 6))
    plt.imshow(wc, interpolation="bilinear")
    plt.title(f"{ethnic_group.capitalize()} Recipes Ingredient Word Cloud (Filtered)", fontsize=16)
    plt.axis("off")
    plt.tight_layout()
    plt.show()


generate_filtered_ethnic_group_wordcloud(df_recipes, ethnic_group="chinese")
```

```python
import pandas as pd
import numpy as np
from collections import Counter, defaultdict
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx

# ========= 参数：可修改 =============
top_n_ethnic = 10  # 选最多的民族
top_m_ing = 30     # 选最常见的食材
# ===================================

# ---------- 筛选Top民族 & 食材 ----------
ethnic_counts = Counter()
ingredient_counts = Counter()

for _, row in df_recipes.iterrows():
    ethnic = row['ethnic_group']
    ings = row['ingredients']
    if pd.notnull(ethnic) and isinstance(ings, list):
        ethnic = ethnic.strip().lower()
        ethnic_counts[ethnic] += 1
        for ing in ings:
            ingredient_counts[ing.strip().lower()] += 1

top_ethnics = set([e for e, _ in ethnic_counts.most_common(top_n_ethnic)])
top_ingredients = set([i for i, _ in ingredient_counts.most_common(top_m_ing)])

# ---------- 构建民族 × 食材矩阵 ----------
data = defaultdict(Counter)
for _, row in df_recipes.iterrows():
    ethnic = row['ethnic_group']
    ings = row['ingredients']
    if pd.notnull(ethnic) and isinstance(ings, list):
        ethnic = ethnic.strip().lower()
        if ethnic not in top_ethnics:
            continue
        for ing in ings:
            ing = ing.strip().lower()
            if ing in top_ingredients:
                data[ethnic][ing] += 1

df_matrix = pd.DataFrame.from_dict(data, orient='index').fillna(0).astype(int)

# ========== 方法一：民族-民族图 ==========
G_ethnic = nx.Graph()
for e1 in df_matrix.index:
    for e2 in df_matrix.index:
        if e1 >= e2:
            continue
        shared_ings = np.logical_and(df_matrix.loc[e1] > 0, df_matrix.loc[e2] > 0).sum()
        if shared_ings > 0:
            G_ethnic.add_edge(e1, e2, weight=shared_ings)

print(f"方法一：民族-民族图共 {G_ethnic.number_of_nodes()} 个民族，{G_ethnic.number_of_edges()} 条边")

# 可视化网络图（可选）
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G_ethnic, seed=42)
edges = G_ethnic.edges(data=True)
weights = [d['weight'] for (_, _, d) in edges]
nx.draw(G_ethnic, pos, with_labels=True, edge_color=weights, edge_cmap=plt.cm.Blues, width=2, node_color='orange')
plt.title("Method 1: Ethnic-Ethnic Graph via Shared Ingredients")
plt.show()

# ========== 方法二：民族相似度热力图 ==========
similarity_matrix = pd.DataFrame(
    cosine_similarity(df_matrix),
    index=df_matrix.index,
    columns=df_matrix.index
)

plt.figure(figsize=(10, 8))
sns.heatmap(similarity_matrix, cmap="YlGnBu", annot=True)
plt.title("Method 2: Cosine Similarity Between Ethnic Groups")
plt.show()
```

## 3D力导向布局图（注：可交互，鼠标拖拽旋转）

对整个二分图B（民族 + 食材）一起做3D力导向布局，然后筛选出民族节点和部分高连接食材节点。力导向布局的输入图是民族和食材的完整二分网络。

节点位置是整体布局考虑了民族与食材节点的相互作用，最终位置会受到食材节点的影响。

```python
import numpy as np
import plotly.graph_objects as go
from tqdm.notebook import tqdm

# 1. 根据边的共现频率筛选
min_edge_weight = 5  # 可调：边共现次数至少多少次
filtered_edges = [(u, v, w) for (u, v), w in co_occurrence_counter.items() if w >= min_edge_weight]

# 2. 统计各个节点（食材）的度数
node_counter = Counter()
for u, v, w in filtered_edges:
    node_counter[u] += 1
    node_counter[v] += 1

# 3. 保留 Top-K 节点（按出现次数）
top_k = 80  # 可调：最多保留多少个食材节点
top_nodes = set([node for node, _ in node_counter.most_common(top_k)])

# 4. 只保留 top_k 中的边
final_edges = [(u, v, w) for u, v, w in filtered_edges if u in top_nodes and v in top_nodes]

# 5. 构建新图
G_filtered = nx.Graph()
for u, v, w in final_edges:
    G_filtered.add_edge(u, v, weight=w)

print(f"Filtered graph: {G_filtered.number_of_nodes()} nodes, {G_filtered.number_of_edges()} edges")


# # 构建 NetworkX 图
# G = nx.Graph()
# for (ing1, ing2), weight in co_occurrence_counter.items():
#     if weight >= 2:  # 可调阈值
#         G.add_edge(ing1, ing2, weight=weight)

# print(f"Number of ingredient nodes: {G.number_of_nodes()}")
# print(f"Number of edges: {G.number_of_edges()}")

# 3D 力导向布局
def spring_layout_3d(G, iterations=50, seed=42, k=0.7):
    np.random.seed(seed)
    pos = {n: np.random.rand(3) for n in G.nodes()}
    for _ in tqdm(range(iterations), desc="3D Layout Progress"):
        pos = nx.spring_layout(G, pos=pos, dim=3, iterations=1, k=k, seed=seed)
    return pos

# 生成 3D 坐标
pos_3d = spring_layout_3d(G_filtered, iterations=30, k=2.5, seed=1)

# 分解为 x, y, z 坐标
xs, ys, zs = [], [], []
node_text = []
for node in G_filtered.nodes():
    x, y, z = pos_3d[node]
    xs.append(x)
    ys.append(y)
    zs.append(z)
    node_text.append(node)

# 绘制边线
edge_x, edge_y, edge_z = [], [], []
for u, v in G_filtered.edges():
    x0, y0, z0 = pos_3d[u]
    x1, y1, z1 = pos_3d[v]
    edge_x += [x0, x1, None]
    edge_y += [y0, y1, None]
    edge_z += [z0, z1, None]

# 使用 Plotly 绘图
fig = go.Figure()

# 添加边
fig.add_trace(go.Scatter3d(
    x=edge_x, y=edge_y, z=edge_z,
    mode='lines',
    line=dict(color='lightgray', width=1),
    hoverinfo='none'
))

# 添加节点
fig.add_trace(go.Scatter3d(
    x=xs, y=ys, z=zs,
    mode='markers+text',
    text=node_text,
    marker=dict(size=6, color='orange'),
    textposition='top center'
))

fig.update_layout(
    title='Ingredient Co-occurrence Network (3D)',
    showlegend=False,
    margin=dict(l=0, r=0, b=0, t=30)
)

fig.show()
```

```python
from collections import Counter
import matplotlib.pyplot as plt

american_books = df_books_copy[
    df_books_copy['categorized_subject'].apply(lambda x: 'American' in x if isinstance(x, list) else False)
]['book_id'].unique()

print(f"包含 'American' 类别的书籍数量: {len(american_books)}")



# 2. 筛选 ethnic_group == 'chinese' 且 book_id 在上述 american_books 中的菜谱
chinese_recipes_american = df_recipes[
    (df_recipes['ethnic_group'].str.lower() == 'chinese') &
    (df_recipes['book_id'].isin(american_books))
].copy()

print(f"'American' 类别书籍中 Chinese 菜谱数量: {chinese_recipes_american.shape[0]}")
print('就那一本书的66个中国菜谱')

# 3. 统计所有食材出现次数，取前10
all_ingredients = [ing for lst in chinese_recipes_american['ingredients'] for ing in lst]
ingredient_counts = Counter(all_ingredients)
top10_ingredients = ['syou','water chestnut','rice flour','peanut oil','bamboo shoot', 'bean sprout']

# 4. 合并书籍年份
chinese_recipes_american = chinese_recipes_american.merge(
    df_books_copy[['book_id', 'date']],
    on='book_id',
    how='left'
).dropna(subset=['date'])

# 确保年份是整数
chinese_recipes_american['date'] = chinese_recipes_american['date'].astype(int)

# 5. 按年份统计每个食材出现次数
top_ingredients = top10_ingredients
data = []
for year, group in chinese_recipes_american.groupby('date'):
    year_ingredients = [ing for lst in group['ingredients'] for ing in lst]
    year_counts = Counter(year_ingredients)
    row = {'year': year}
    for ing in top_ingredients:
        row[ing] = year_counts.get(ing, 0)
    data.append(row)

df_yearly = pd.DataFrame(data).sort_values('year').set_index('year')

# 6. 绘图展示
plt.figure(figsize=(8,6))
for ing in top_ingredients:
    plt.plot(df_yearly.index, df_yearly[ing], marker='o', label=ing)
plt.xlabel('year')
plt.ylabel('frequency')
# plt.title('书籍类别为American的中国菜谱中前十食材按年份出现次数')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
